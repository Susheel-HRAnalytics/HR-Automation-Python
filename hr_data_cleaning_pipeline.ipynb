```python
import pandas as pd
import numpy as np
```

```python
try:
    df = pd.read_csv('hr_recruitment_1000_records.csv')
    print(" Dataset loaded successfully!")
except FileNotFoundError:
    # If file isn't uploaded yet, we create a small sample to show it works
    print(" File not found. Creating a sample dataset to demonstrate...")
    data = {
        'Candidate_ID': ['C001', 'C002', 'C003'],
        'Candidate_Name': ['Amit Sharma', 'Priya Patel', 'Rohan Das'],
        'Recruitment_Source': ['LinkedIn', 'Indeed', 'Referral'],
        'Application_Date': ['2026-01-05', '2026-01-07', '2026-01-10'],
        'Hire_Date': ['2026-02-10', np.nan, '2026-01-25'],
        'Status': ['Hired', 'Rejected', 'Hired']
    }
    df = pd.DataFrame(data)
```

```python
# 2. Data Cleaning: Convert date columns to standard datetime formats
df['Application_Date'] = pd.to_datetime(df['Application_Date'])
df['Hire_Date'] = pd.to_datetime(df['Hire_Date'])
```

```python
# 3. Data Transformation: Calculate 'Time-to-Hire' safely for hired candidates only
df['Time_To_Hire_Days'] = (df['Hire_Date'] - df['Application_Date']).dt.days
```

```python
# Fill missing values for rejected candidates gracefully
df['Time_To_Hire_Days'] = df['Time_To_Hire_Days'].fillna('N/A')
```

```python
# 4. View the cleaned data overview
print("\n Cleaned Data Summary (First few rows):")
print(df.head())
```

```python
# 5. Export the clean data back to a new CSV file
df.to_csv('cleaned_hr_recruitment_data.csv', index=False)
print("\n Process complete! 'cleaned_hr_recruitment_data.csv' has been generated.")
```
